# Git Version Control
[Git](https://git-scm.com/install/)
- Tracks changes you make over time so you can see the full history and revert mistakes.
- Lets you create separate “branches” to work on without ruining your primary content that you can then merge back when ready.
- Repositories contain the whole project. You can work offline and sync later.
- Git != Github

## Post-Install

First verify that git is installed and note which version you are using:
```
$ which git
$ git --version
```

Git has its own configuration that it uses as you work. To view the current config:
```sh
$ git config --list
```

There are global and local settings. Global is used everywhere, but values can be overridden by the local configuration:
```
$ git config --list --global
$ git config --list --local
```
To set a value, use the name of the configuration and the new value. For example, to set your identity across all your work:
```sh
$ git config --global user.name "Jane Doe"
$ git config --global user.email johndoe@example.com
```

If you are working on a project that requires a different email, you could do:
```sh
$ git config --local user.name "John Deer"
$ git config --local user.email johndeer@otherexample.com
```

## Creating a Repository

With Git configured, let's create a repository!

Create a directory in your home folder named `planets` and `cd` into that directory:

```sh
$ cd ~
$ mkdir planets
$ cd planets
```

Next, make the `planets` directory a Git repository:
```sh
$ git init
```

**Note:** A git repository includes all the subdirectories and all the files in those subdirectories. There is no need to create separate repositories nested within the `planets` repository.

Run `ls` to verify that a folder named `.git` now exists in your `planets` directory. If you don't see it, try adding the `-a` (all) flag to your command. e.g. `ls -a`

Git uses this special subdirectory to keep track of all your changes to all files and sub-directories located within the project. If you don't want this directory to be seen as a Git repository, you can simply delete the `.git` directory.

**Warning:** Depending on how you delete the folder, this action may not be recoverable.

We can check that everything is set up correctly by asking Git to tell us the status of our project:
```sh
$ git status
```

## Making Changes

In your `planets` repository, create a file named `mars.txt` and add some notes and comments about the Red Planet using vim (neo or classic)
```sh
$ vim mars.txt
```

Save the file, exit vim, and check the status of your repository. Git should tell us that it’s noticed the new file:
```sh
$ git status
```

Your output should read something like:
```sh
On branch main

No commits yet

Untracked files:
   (use "git add <file>..." to include in what will be committed)
	mars.txt

nothing added to commit but untracked files present (use "git add" to track)
```

The “untracked files” message means that there’s a file in the directory that Git isn’t keeping track of. We can tell Git to track a file using `git add`:
```sh
$ git add mars.txt
```

Then, let's check that this new file is accounted for:
```sh
$ git status
```
and you should now see something like:
```sh
On branch main

No commits yet

Changes to be committed:
  (use "git rm --cached <file>..." to unstage)

	new file:   mars.txt
```

Git now knows that it’s supposed to keep track of `mars.txt`, but it hasn’t recorded these changes as a commit yet. To get it to do that, we need to run one more command:
```sh
$ git commit -m "Added information about mars"
 ```
 Output should read something like:
```sh
[main (root-commit) de5825f] Added information about mars
 1 file changed, 1 insertion(+)
 create mode 100644 mars.txt
```

**Note:** We use the `-m` flag (for “message”) to record a short, descriptive, and specific comment that will help us remember later on what we did and why. If we run `git commit` without the `-m` option, Git will launch your default text editor so that we can write a longer message. Generally, we want to keep commits short and sweet, less than 50 characters: ex: "Fixed typos in functions.py". Capture the work, not the minutia.

When we run git commit, Git takes everything we have told it to track and saves the delta or changes inside the special `.git` directory. This permanent copy is called a [commit (or revision)](#glossary) and it is giving a hashed identifier i.e. `de5825f`. Your identifier may vary.

After making our commit, `git status` should now show something similar to:
```sh
On branch main
nothing to commit, working directory clean
```

This tells us that everything is up to date. To see our recent changes, we can check the log:

```sh
commit de5825f9f36eab3dccc1f849c9fd8cf3607c7d5a (HEAD -> main)
Author: Stryder <stryder@strangedreams.com>
Date:   Tue Aug 18 00:00:17 2026 -0400

    Added information about mars
```

Let's open `mars.txt` again and add some new information about Mar's moons:
```bash
$ vim mars.txt
```

Running `git status` now should show us git knows that `mars.txt` has been modified:
```sh
On branch main
Changes not staged for commit:
  (use "git add <file>..." to update what will be committed)
  (use "git restore <file>..." to discard changes in working directory)
        modified:   mars.txt

no changes added to commit (use "git add" and/or "git commit -a")
```

The last line is the key phrase: `“no changes added to commit”`. We have changed this file, but we haven’t told Git we will want to track those changes (which we do with `git add`) nor have we saved them (which we do with `git commit`). So let’s do that now. It is good practice to always review our changes before saving them. We do this using `git diff`. This shows us the differences between the current state of the file and the most recently saved version:

```bash
$ git diff
```
Output:
```bash
diff --git a/mars.txt b/mars.txt
index de5825f..315bf3a 100644
--- a/mars.txt
+++ b/mars.txt
@@ -1 +1,2 @@
Cold and dry, but everything is my favorite color.
The two moons may be a problem for the Mickey Mousoids...
```

The output is cryptic because it is actually a series of commands that we can break down into pieces:

1. The first line tells us that Git is producing output similar to the Unix diff command comparing the old and new versions of the file.
2. The second line tells exactly which versions of the file Git is comparing; df0654a and 315bf3a are unique computer-generated labels for those versions.
3. The third and fourth lines once again show the name of the file being changed.
4. The remaining lines are the most interesting, they show us the actual differences and the lines on which they occur. In particular, the + marker in the first column shows where we added a line.

After reviewing our changes, it’s time to commit them:
```bash
$ git add mars.txt
$ git commit -m "Added information about Mar's moons"
$ git status
```

## SSH (Secure SHell)
- Secure protocol that provides secure communication between two computers using keys.
- Asymmetric encryption uses two keys: one public and one private
- The private key is kept on your local machine and **must** remain confidential
- The public key is shared with servers or people that you want to be able to verify you

During SSH connection, the client proves its identity via private key through a challenge‑response process. The challenger sends a packet with a special message encrypted with the public key that can only be decrypted with the private key. The challenged computer then decrypts the message, and sends it back encrypted with the challengers public key, that only the challenge can decode with its private key, thus establishing each client's identity.

- SSH keys *must be unique*.
- **ONE PER MACHINE PER IDENTITY**
- **DO NOT RE-USE KEYS**
- **DO NOT SHARE PRIVATE KEYS**
- ## DO NOT SHARE PRIVATE KEYS

## SSH Setup

To setup an SSH identity, first check if any existing SSH keys already exist.

```bash
ls ~/.ssh
```

If you see a pair of files, one without a prefix and one ending with *`.pub`*, that indicates a key pair. For example, you might see
*`id_ed25519`* and *`id_ed25519.pub`*.

If these files exist in your ~/.ssh folder, you can jump to the [Adding SSH key](#AddSSH) to your GitHub account section of this tutorial.

If you do not have any SSH key pairs or need to generate new one:

1. In your terminal, run `ssh-keygen -t ed25519 -C "youremail@example.org"`

Expected output:
```
Generating public/private ed25519 key pair.
Enter file in which to save the key (/pathto/user/home/.ssh/id_ed25519)
```

(The path in parenthesis is the default. Accept it unless you have specific alternate instructions.)

Next, you will be prompted for a passphrase. Going forward, this is the passphrase you will use when you have to authenticate this identity. If you don't enter a passphrase and just hit enter, you will not be asked for a password going forward. This password is **not** recoverable!

```
> Enter passphrase (empty for no passphrase): [Type a passphrase]
> Enter same passphrase again: [Type passphrase again]
```

You should then see an output similar to this:
```
Your identification has been saved in /home/daxfeliz/.ssh/id_ed25519
Your public key has been saved in /home/daxfeliz/.ssh/id_ed25519.pub
The key fingerprint is:
SHA256:AuErG+8I8YUkRbNn1iNiGB/T3P6p4oWtmHA821i3bPO dfeliz@amnh.org
The key's randomart image is:
+--[ED25519 256]--+
|    ..o.o        |
|.o . o o o       |
|o O o . o .    . |
|.* * o . .  . o  |
|*++ + . S  . E   |
|**oo   .    . .  |
|*++.    .    o   |
|o=o.     . .o.   |
|+o.       ..o..  |
+----[SHA256]-----+
```

Congratulations! Your key has been created!

## Adding the SSH key to the ssh-agent process
If you are on Linux, you will need to add the generated key to a process called `ssh-agent`.

```sh
eval "$(ssh-agent -s)"
ssh-add ~/.ssh/id_ed25519
```

On MacOS, you can add your ssh key to the apple key chain:
```sh
ssh-add --apple-use-keychain ~/.ssh/id_ed25519
```

## Github and SSH
Github now requires SSH to interact programmatically with your repositories, so we have to let Github know what our machines public key is.

### Adding SSH keys to your GitHub account

* Login to GitHub
* Click on your profile in upper right corner and select **Settings**
* On the left sidebar, click on **SSH and GPG keys**.
* Click on **New SSH key**.
* Assign a Title for this key. This is just a label for you to remember which machine this key is associated with.
* Paste the key from the `.pub` file that was created when you generated your .ssh key
* Click **Add SSH key**. You might be prompted to type your GitHub password.

Next, you can test your connection from your Bash environment.

In your Terminal, type the following exactly:
```sh
ssh -T git@github.com
```

You might see a similar warning as follows (including the public key as shown below):
```
The authenticity of host 'github.com (140.82.114.3)' can't be established.
RSA key fingerprint is SHA256:nThbg6kXUpJWGl7E1IGOCspRomTxdCARLviKw6E5SY8.
Are you sure you want to continue connecting (yes/no/[fingerprint])?
```

Type **`yes`** to continue.

The next warning should list your account name:
```
Warning: Permanently added 'github.com' (RSA) to the list of known hosts.
Hi <name>! You've successfully authenticated, but GitHub does not provide shell access.
```

Voila! Your identity key is added and we can now push our work up to Github!

## Using Github

Github *is not Git*.  Github is a site that hosts repositories so that they can be shared with other users.

Let's add our `planets` repository to our Github account!

1. Open [Github](https://github.com) and login if you need to.
2. Select 'Repositories' from the menu across the top or click on your profile and select 'Repositories'
3. Click 'New'
4. Fill out the form (info below)
5. Click 'Create Repository'

**Repository Form Fields**
- *Repository Name:*  Used to identify your repo in urls etc...Usually the same as your project name
- *Description:* A brief summary of your project for other users (and you)
- *Visibility:* Who can see/access your repository
- *Add README:* A markdown file that will be used for your repository landing page on github.com
- *Add .gitignore:* A file in your repository that git uses to automatically 'ignore' certain files and folders
- *Add license:* A markdown file that you can use to provide licensing information about your project/data

Now that your new repository on github has been created, we can connect it to our already existing repository.
```sh
$ git remote add origin git@github.com:<username>/planets.git
$ git branch -M main
$ git push -u origin main
```

Refresh your repository page and review your code!
