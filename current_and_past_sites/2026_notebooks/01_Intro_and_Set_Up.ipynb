# The Terminal
A Terminal Emulator or 'Command Line' is an application that lets you issue commands directly to your computer.
On new computers, it's just called 'Terminal' or 'Command Prompt' and can found from your Application or Start Menu.
The Terminal program that comes with your computer by default is usually pretty barebones, but you can download and
install other emulators that offer additional functionality like session management and integrated graphics.

It's highly recommended to pin your preferred Terminal Emulator somewhere since you'll likely be using it a lot.

## Basics
Your Terminal uses a *Shell* to interpret the commands you enter and displays the results. They come in many varieties, but a
few of the most common you'll come across are [Bash](	https://www.gnu.org/software/bash/), [Zsh](https://www.zsh.org/
)("Zee-Shell"), and [Fsh](https://fishshell.com/)("Fish"). Of these, Bash is default on most Linux systems and Zsh is the default on OSX. Sounds like a lot, but good news: They all work almost exactly the same.  Shell commands all have the same basic format: the command name followed by a list of parameters, flags, or arguments.

> **Exercise:**
> 1. Open your terminal now and execute the command `pwd`, what do you see?
> 2. Now execute the command `ls`, what do you see?
> 3. Run `touch foo.txt` and then `ls` again, what changed?


The `pwd` command stands for "Present Working Directory" and it shows you the current directory your shell is working in.
`ls` is short for "list" and it will show you the contents of the current directory. The `touch` command will create an empty file with the name you provided.

Other basic commands you should know:
- `mkdir <directory_name>` - makes a directory with the given name
- `cd <directory_name>` - used to *change* your current directory
- `rm <file_name>` - *removes* or deletes a file
- `rmdir <directory_name>` - *removes* or deletes a directory
- `mv <file1> <file2>` - *moves* `file1` to `file2`.
- `cp <file1> <file2>` - *copies* `file1` to `file2`.
- `open <file1>` - *opens* the target file with the default application
- `which <command>` - tells you *which* command will be run

Shell Commands all follow the same format: A command name followed by a series of *flags*, *options*, or *parameters*. These arguments can change or modify behavior in someway. `ls` for example works using the current directory by default, but you can also pass the name of a directory. `mkdir` similarly works in the current directory. If you pass multiple names, it will create each of those directories. If you try to create nested directories, it will fail unless you pass the `-p` (parents) argument since those child directories don't exist. `rmdir` won't work on directories with content unless you use the `-r` (recursive) argument.

To learn more about the various arguments and parameters a command uses, you can always try to pass the `--help` flag or use the `man` command followed by the name of the command you want to learn about. e.g. `man mkdir`

## About Paths, Filenames, and Extensions
A path is a series of names separated by forward slashes (`/`) that tell the computer where something is. On Linux and MacOS (*nix systems), a slash on its own represents the *root* directory of your file system. On Windows, the root is `c:\`. *nix and Windows disagree on how exactly a path should be expressed. Windows is generally **not** case sensitive, but *nix systems are. This means that where Windows would see `my_data.txt` and `My_Data.txt` as the same file, on *nix, they would be two disctinct files.

The `cd` command supports a few special targets:
 - `~` - takes you back to your *home* directory
 - `..` - moves you 'up' on directory
 - `.` - references the current directory you're in

Extensions are the 'dot something' suffix you see on a file, like `.txt`, `.pdf` or `.jpg`. These are just hints to help you and the computer identify what type of content *should* be in the file. Your computer uses the extension to pick which application to use to open it. You can give a file any extension you want and there are a myriad of them out there.

## Editing Files In Your Terminal
[Neovim](https://neovim.io)

- Fork of the classic Vim editor. Same editor, more module, bit cleaner.
- Add async support, Lua scripting, a modern plugin API, and simplified extension development.
- Barebones text editor or robust IDE. Your pick.
- Recommended Plugin: [AstroVim](https://github.com/AstroNvim/AstroNvim)

Once installed, you can launch neovim from the terminal by typing `nvim` (may require you to restart your terminal)

Vim has two basic modes: *Normal* and *Insert*. There are other modes that we'll cover indepth later.

Typical Workflow:
* `nvim <file>` will open that file for editing
* To make edits, press `"i"` once to enter *insert mode*.
* Press `"esc"` to exit insert mode and go into *normal mode*
* Type `:` in normal mode to go to command mode.
* Type `:w` to *write* your changes and `:q` to *quit*.
* Commands can be combined. i.e. `:wq` will *write* and then *quit*.
* Use `!` to force a command. i.e. `:q!`.
* To overwrite the file and force quit to exit the vim text editor in one step, type in `:wq!`

## Editing your shell environment

When you login to your computer and start up your shell, it gets information about its environment and commands from a 'run-command' file. By convention, this file lives in your *home directory* and is named `.<shell>rc`.  So if you are using a *bash* shell, you'll have a `.bashrc` file and a `.zshrc` file if you're using zsh.

1. Open your terminal
2. Type `ls -a`. Tye `-a` shows *all* files. Files and directories that start with a `.` are *hidden* by default.
3. Identify your rc file and open it with your terminal text editor

The rc file is a useful way to customize your environment and tell your shell where it can find other commands, For example, you could `alias` vi to always launch neovim by adding the following line:

`alias vi=nvim`

Write that to the bottom of your rc file, save it, exit and then run
`source .<shell>rc`. (`source` is a command that will read your rc file into the current shell environment)

Run `vi` again and notice how it launches neovim instead. We'll go over more commands to add in future lessons.

#### \*A Note About Anaconda!

Anaconda is a nominally free, special distribution of Python that also acts as a package and environment  manager with a focus on scientific libraries and binaries. However, in 2024, Anaconda clarified its Terms of Service so that organizations with $\ge$ 200 employees or contractors must buy a paid Business license to use Anaconda’s repositories (the default “defaults” channel many conda installs use); this threshold applies to companies, governments, and non-profits, with separate carve-outs for accredited educational use (like universities) in course curricula. Practically, if your conda/Miniconda setup downloads from Anaconda’s repos inside a large organization, you’re expected to pay—or switch to a community source like conda-forge via Miniforge. More details about the changes in licensing can be found  [__here__](https://www.datacamp.com/blog/navigating-anaconda-licensing).

To address this, institutions like Flatiron Institute Center for Computational Astrophysics (a non-profit organization) implemented a firewall blocking downloads from conda's default channel. You can use default `conda` as a package manager but you'll have to download packages from conda-forge if using conda while at CCA.

Miniforge, mentioned above, is a mini-installer of anaconda and defaults to installing packages from the (free!) community-based conda-forge channel and is organization independent: university vs non-profit.

Should you want to use conda now or in the future, you can always use the instructions found in [anaconda.md](docs/anaconda.md).
